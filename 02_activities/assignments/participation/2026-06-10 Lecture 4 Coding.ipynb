```python
y2_sd=np.std(y2)
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color='red')
#how to customize error bar appearance
ax.errorbar(x, y2,
            yerr = y2_sd,
            fmt="none",
            ecolor="indigo",
            elinewidth=4,
            capsize=6,
            capthick=4)
```

```python
#load libraries
from PIL import Image
import requests
from io import BytesIO

#get image from the internet
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

response=requests.get(url,headers=headers)
image_file=BytesIO(response.content)
image=Image.open(image_file) #to get the file we extracted
image #to show the image on our end
```

```python
#add an image to a plot
fig, ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color="red")
#try to get a box, by specifying coordinates, then put blank box into the figure so it has a black box that acts as a container
ax_image = fig.add_axes([0.10, # x coordinate (ON FIGURE, NOT AXES)
                         0.11, # y coordinate (ON FIGURE, NOT AXES)
                         0.15, # image width
                         .35]) # image height
ax_image.imshow(image) #add image to plot, note that it automatically adjusts the size of the box and tries to place it somewhere based on the container we just created
ax_image.axis('off') #to remove the axis for the container (previously were black but adding this removes them)

#code to save the figure to your computer (best to do this in the same chunk of code that you created the image)
path = 'C:/Users/ziral/Documents/MSc_PhD Nutritional Science/Data Science Institute Certificate/Assignments/visualization/02_activities/assignments'
filename = '/Lecture4_Luffyfig.png'
plt.savefig(path+filename, dpi=300) #dpi is to specify the resolution
```
